```python

!pip -q install networkx scipy

import json
import re
import networkx as nx
import matplotlib.pyplot as plt
from pathlib import Path
from collections import Counter

DATA_DIR = Path("/content/rands_dynamic")

# Load one actual report
report_path = next(DATA_DIR.rglob("*.json"))

with open(report_path, "r", encoding="utf-8") as f:
    report = json.load(f)

G = nx.Graph()

sample_id = report_path.stem
root = f"SAMPLE:{sample_id}"

G.add_node(root, node_type="sample")

# Categories represented as graph entities
categories = {
    "created_process": "process",
    "killed_process": "process",
    "processes_tree": "process_tree",
    "created_files": "file",
    "accessed_files": "file",
    "deleted_files": "file",
    "changed_files": "file",
    "accessed_registry_keys": "registry",
    "created_registry_keys": "registry",
    "network_traffic_ips": "network",
    "network_dns_lookups": "dns",
    "created_mutexes": "mutex",
    "accessed_mutexes": "mutex",
    "loaded_modules": "module",
    "executed_commands": "command",
    "critical_api_calls": "api"
}

# Add observed entities and report membership edges
for category, node_type in categories.items():

    values = report.get(category, [])

    if not isinstance(values, list):
        continue

    counts = Counter(str(x).strip() for x in values if str(x).strip())

    for value, frequency in counts.items():

        node_id = f"{node_type}:{value}"

        G.add_node(node_id, node_type=node_type)

        G.add_edge(
            root,
            node_id,
            relation="observed_in_report",
            weight=frequency
        )

# Extract network IP addresses
ip_pattern = re.compile(
    r"\b(?:\d{1,3}\.){3}\d{1,3}\b"
)

network_ips = set()
dns_ip_links = []

for item in report.get("network_traffic_ips", []):

    parts = str(item).split(" - ")

    if len(parts) >= 3:
        protocol = parts[0].strip().upper()
        ip = parts[1].strip()
        port = parts[2].strip()

        if ip_pattern.fullmatch(ip):
            network_ips.add(ip)

            ip_node = f"network_ip:{ip}"
            G.add_node(ip_node, node_type="network_ip")

            G.nodes[ip_node]["protocol"] = protocol
            G.nodes[ip_node]["port"] = port

for item in report.get("network_dns_lookups", []):

    text = str(item)

    ips = set(ip_pattern.findall(text))
    domain_match = re.search(
        r"-\s*([A-Za-z0-9._-]+\.[A-Za-z]{2,})\s*$",
        text
    )

    if domain_match:
        domain = domain_match.group(1)
        domain_node = f"domain:{domain}"

        G.add_node(domain_node, node_type="domain")

        for ip in ips.intersection(network_ips):
            G.add_edge(
                domain_node,
                f"network_ip:{ip}",
                relation="shared_dns_ip",
                weight=1
            )

            dns_ip_links.append((domain, ip))

# Connect process and command observations only on basename match
import ntpath

def executable_name(value):
    value = str(value).strip().strip('"')
    first = value.split()[0] if value else ""
    return ntpath.basename(first).lower()

process_nodes = {}

for value in report.get("created_process", []):
    name = executable_name(value)
    if name:
        process_nodes.setdefault(name, []).append(
            f"process:{str(value).strip()}"
        )

matched_process_commands = 0

for command in report.get("executed_commands", []):
    name = executable_name(command)

    if name and name in process_nodes:
        command_node = f"command:{str(command).strip()}"

        for process_node in process_nodes[name]:
            G.add_edge(
                process_node,
                command_node,
                relation="executable_name_match",
                weight=1
            )
            matched_process_commands += 1

# Calculate graph features
degree = dict(G.degree())
weighted_degree = dict(G.degree(weight="weight"))

core = nx.core_number(G) if G.number_of_edges() else {}

try:
    eigen = nx.eigenvector_centrality(
        G, max_iter=1000, weight="weight"
    )
except Exception:
    eigen = {n: 0.0 for n in G.nodes()}

# Basic graph measurements
print("SAMPLE:", sample_id)
print("\nGRAPH SUMMARY")
print("Nodes:", G.number_of_nodes())
print("Edges:", G.number_of_edges())
print("Density:", nx.density(G))
print("Connected components:", nx.number_connected_components(G))
print("Average degree:", sum(degree.values()) / max(len(degree), 1))
print("Maximum k-shell:", max(core.values(), default=0))
print("Mean k-shell:", sum(core.values()) / max(len(core), 1))
print("Maximum eigenvector centrality:", max(eigen.values(), default=0))
print("Mean eigenvector centrality:", sum(eigen.values()) / max(len(eigen), 1))

print("\nNETWORK FEATURES")
print("Unique network IPs:", len(network_ips))
print("DNS-to-IP matches:", len(dns_ip_links))
print("Matched process-command associations:", matched_process_commands)

# Protocol and port features
protocol_counts = Counter()
ports = set()

for item in report.get("network_traffic_ips", []):
    parts = str(item).split(" - ")

    if len(parts) >= 3:
        protocol_counts[parts[0].strip().upper()] += 1
        ports.add(parts[2].strip())

print("Protocol counts:", dict(protocol_counts))
print("Unique ports:", len(ports))

# Draw graph
plt.figure(figsize=(14, 10))

pos = nx.spring_layout(G, seed=42, k=0.5)

nx.draw_networkx(
    G,
    pos,
    node_size=35,
    with_labels=False,
    width=0.5
)

plt.title("RanDS Sample Behaviour Graph")
plt.axis("off")
plt.show()
```

```python

import networkx as nx
import matplotlib.pyplot as plt
import json
import re
import ntpath
from pathlib import Path

DATA_DIR = Path("/content/rands_dynamic")
report_path = next(DATA_DIR.rglob("*.json"))

with open(report_path, "r", encoding="utf-8") as f:
    report = json.load(f)

G_clean = nx.Graph()

# Extract IP addresses
ip_pattern = re.compile(r"\b(?:\d{1,3}\.){3}\d{1,3}\b")

network_ips = set()

for item in report.get("network_traffic_ips", []):
    parts = str(item).split(" - ")

    if len(parts) >= 3:
        protocol, ip, port = parts[:3]

        if ip_pattern.fullmatch(ip.strip()):
            ip = ip.strip()
            network_ips.add(ip)

            G_clean.add_node(
                "IP:" + ip,
                node_type="network"
            )

# DNS-to-IP relationships
for item in report.get("network_dns_lookups", []):
    text = str(item)
    ips = set(ip_pattern.findall(text))

    domain_match = re.search(
        r"-\s*([A-Za-z0-9._-]+\.[A-Za-z]{2,})\s*$",
        text
    )

    if domain_match:
        domain = domain_match.group(1)
        domain_node = "DNS:" + domain

        G_clean.add_node(domain_node, node_type="dns")

        for ip in ips.intersection(network_ips):
            G_clean.add_edge(
                domain_node,
                "IP:" + ip,
                relation="dns_ip_match"
            )

# Process-command relationships based on executable-name matches
def exe_name(value):
    value = str(value).strip().strip('"')
    return ntpath.basename(value.split()[0]).lower() if value else ""

process_map = {}

for process in report.get("created_process", []):
    name = exe_name(process)

    if name:
        process_map.setdefault(name, []).append(str(process))

for command in report.get("executed_commands", []):
    name = exe_name(command)

    if name in process_map:
        command_node = "CMD:" + str(command)

        G_clean.add_node(command_node, node_type="command")

        for process in process_map[name]:
            process_node = "PROC:" + process
            G_clean.add_node(process_node, node_type="process")

            G_clean.add_edge(
                process_node,
                command_node,
                relation="executable_name_match"
            )

# Remove isolated nodes for clearer relationship visualisation
G_clean.remove_nodes_from(list(nx.isolates(G_clean)))

print("Evidence-linked nodes:", G_clean.number_of_nodes())
print("Evidence-linked edges:", G_clean.number_of_edges())
print("Connected components:",
      nx.number_connected_components(G_clean))

if G_clean.number_of_nodes() > 0:
    pos = nx.spring_layout(G_clean, seed=42)

    plt.figure(figsize=(14, 9))
    nx.draw(
        G_clean,
        pos,
        with_labels=True,
        node_size=500,
        font_size=7,
        width=1.2
    )

    plt.title("Evidence-linked RanDS Behaviour Graph")
    plt.axis("off")
    plt.tight_layout()

    plt.savefig(
        "/content/RanDS_Evidence_Graph.png",
        dpi=200,
        bbox_inches="tight"
    )
    plt.show()

    print("Image saved: /content/RanDS_Evidence_Graph.png")
else:
    print("No evidence-linked relationships found.")
```

```python

import networkx as nx
import numpy as np
import pandas as pd
from scipy.sparse import csr_matrix

# --------------------------------------------------
# INPUT: evidence-linked graph from your previous cell
# --------------------------------------------------
G = G_clean.copy()

if G.number_of_nodes() == 0:
    raise ValueError("G_clean is empty. Build the evidence-linked graph first.")

# These calculations use an undirected structural graph.
# Preserve your original graph separately if it is directed.
if G.is_directed():
    G = G.to_undirected()

# Remove self-loops; they can distort structural metrics.
G.remove_edges_from(nx.selfloop_edges(G))

# --------------------------------------------------
# A. K-SHELL / CORE NUMBER
# --------------------------------------------------
# Standard k-core decomposition works across disconnected
# components without connecting them.
if G.number_of_edges() > 0:
    core_numbers = nx.core_number(G)
else:
    core_numbers = {node: 0 for node in G.nodes()}

# --------------------------------------------------
# B. EIGENVECTOR CENTRALITY PER COMPONENT
# --------------------------------------------------
eigenvector = {node: 0.0 for node in G.nodes()}
component_id = {}
component_sizes = {}

components = sorted(
    nx.connected_components(G),
    key=len,
    reverse=True
)

for cid, nodes in enumerate(components):
    nodes = list(nodes)
    subgraph = G.subgraph(nodes)

    for node in nodes:
        component_id[node] = cid
        component_sizes[node] = len(nodes)

    # A singleton has no meaningful interaction-based importance.
    if len(nodes) == 1 or subgraph.number_of_edges() == 0:
        continue

    try:
        scores = nx.eigenvector_centrality(
            subgraph,
            max_iter=5000,
            tol=1e-10
        )
    except nx.PowerIterationFailedConvergence:
        # Fallback for difficult convergence cases.
        scores = nx.eigenvector_centrality_numpy(subgraph)

    eigenvector.update(scores)

# --------------------------------------------------
# C. KRYLOV-BASED FEATURES PER COMPONENT
# --------------------------------------------------
# For each component:
# A_norm = D^(-1/2) A D^(-1/2)
# b = normalized all-ones starting vector
#
# Krylov vectors:
# b, A_norm*b, A_norm^2*b, A_norm^3*b
#
# These describe structural reach/influence over successive
# graph steps under this mathematical operator.

krylov_node_values = {
    node: {"krylov_1": 0.0, "krylov_2": 0.0, "krylov_3": 0.0}
    for node in G.nodes()
}

component_krylov = []

for cid, nodes in enumerate(components):
    nodes = list(nodes)
    subgraph = G.subgraph(nodes)
    n = len(nodes)

    if n == 1 or subgraph.number_of_edges() == 0:
        component_krylov.append({
            "component_id": cid,
            "nodes": n,
            "edges": subgraph.number_of_edges(),
            "krylov_norm_1": 0.0,
            "krylov_norm_2": 0.0,
            "krylov_norm_3": 0.0
        })
        continue

    A = nx.to_scipy_sparse_array(
        subgraph,
        nodelist=nodes,
        weight=None,
        dtype=float,
        format="csr"
    )
    A = csr_matrix(A)

    degrees = np.asarray(A.sum(axis=1)).ravel()
    inv_sqrt_degree = np.zeros_like(degrees, dtype=float)
    nonzero = degrees > 0
    inv_sqrt_degree[nonzero] = 1.0 / np.sqrt(degrees[nonzero])

    D_inv = csr_matrix(
        (inv_sqrt_degree, (np.arange(n), np.arange(n))),
        shape=(n, n)
    )

    A_norm = D_inv @ A @ D_inv

    # Deterministic starting vector
    b = np.ones(n, dtype=float) / np.sqrt(n)

    v1 = A_norm @ b
    v2 = A_norm @ v1
    v3 = A_norm @ v2

    vectors = [v1, v2, v3]

    for i, vector in enumerate(vectors, start=1):
        for node, value in zip(nodes, np.asarray(vector).ravel()):
            krylov_node_values[node][f"krylov_{i}"] = float(value)

    component_krylov.append({
        "component_id": cid,
        "nodes": n,
        "edges": subgraph.number_of_edges(),
        "krylov_norm_1": float(np.linalg.norm(v1)),
        "krylov_norm_2": float(np.linalg.norm(v2)),
        "krylov_norm_3": float(np.linalg.norm(v3))
    })

# --------------------------------------------------
# D. NODE-LEVEL FEATURE TABLE
# --------------------------------------------------
node_rows = []

for node in G.nodes():
    row = {
        "node": str(node),
        "degree": int(G.degree(node)),
        "k_shell": int(core_numbers[node]),
        "eigenvector_centrality": float(eigenvector[node]),
        "component_id": int(component_id[node]),
        "component_size": int(component_sizes[node])
    }
    row.update(krylov_node_values[node])
    node_rows.append(row)

node_features = pd.DataFrame(node_rows)

# --------------------------------------------------
# E. SAMPLE-LEVEL GRAPH FEATURE SUMMARY
# --------------------------------------------------
# Node-weighted summaries retain the contribution of all
# components, while component count/size describe fragmentation.

graph_features = {
    "graph_nodes": G.number_of_nodes(),
    "graph_edges": G.number_of_edges(),
    "graph_components": len(components),
    "graph_density": nx.density(G),
    "graph_average_degree": (
        float(np.mean([d for _, d in G.degree()]))
        if G.number_of_nodes() else 0.0
    ),
    "graph_max_degree": (
        int(max(dict(G.degree()).values()))
        if G.number_of_nodes() else 0
    ),
    "graph_max_k_shell": int(node_features["k_shell"].max()),
    "graph_mean_k_shell": float(node_features["k_shell"].mean()),
    "graph_max_eigenvector": float(
        node_features["eigenvector_centrality"].max()
    ),
    "graph_mean_eigenvector": float(
        node_features["eigenvector_centrality"].mean()
    ),
    "graph_mean_krylov_1": float(node_features["krylov_1"].mean()),
    "graph_mean_krylov_2": float(node_features["krylov_2"].mean()),
    "graph_mean_krylov_3": float(node_features["krylov_3"].mean()),
    "graph_max_krylov_1": float(node_features["krylov_1"].max()),
    "graph_max_krylov_2": float(node_features["krylov_2"].max()),
    "graph_max_krylov_3": float(node_features["krylov_3"].max()),
    "largest_component_nodes": int(max(map(len, components))),
}

graph_feature_df = pd.DataFrame([graph_features])
component_feature_df = pd.DataFrame(component_krylov)

# Save outputs
node_features.to_csv(
    "/content/RanDS_Node_Graph_Features.csv", index=False
)
graph_feature_df.to_csv(
    "/content/RanDS_Sample_Graph_Features.csv", index=False
)
component_feature_df.to_csv(
    "/content/RanDS_Component_Krylov_Features.csv", index=False
)

with pd.ExcelWriter(
    "/content/RanDS_Graph_Features.xlsx",
    engine="openpyxl"
) as writer:
    node_features.to_excel(writer, sheet_name="Node_Features", index=False)
    graph_feature_df.to_excel(writer, sheet_name="Sample_Summary", index=False)
    component_feature_df.to_excel(writer, sheet_name="Components", index=False)

print("GRAPH FEATURE EXTRACTION COMPLETE")
print("\nSample-level features:")
display(graph_feature_df.T)

print("\nComponent-level Krylov features:")
display(component_feature_df)

print("\nNode-level preview:")
display(node_features.head(10))

print("\nSaved:")
print("/content/RanDS_Graph_Features.xlsx")
```

```python

# ============================================================
# RanDS: BATCH BEHAVIOURAL + NETWORK + EVIDENCE GRAPH FEATURES
# ============================================================

import os
import re
import json
import glob
import ipaddress
import warnings
from collections import Counter

import numpy as np
import pandas as pd
import networkx as nx
from scipy.sparse import csr_matrix
from tqdm.auto import tqdm

warnings.filterwarnings("ignore", category=RuntimeWarning)

# ------------------------------------------------------------
# 1. PATHS AND SETTINGS
# ------------------------------------------------------------

REPORT_DIR = "/content/rands_dynamic/dataset"
METADATA_FILE = "/content/RanDS_Metadata_Label_Clean.csv"

OUTPUT_FILE = "/content/RanDS_All_Dynamic_GTP_Features.csv"
ERROR_FILE = "/content/RanDS_Extraction_Errors.csv"

CHECKPOINT_SIZE = 500

# ------------------------------------------------------------
# 2. LOAD METADATA AND CREATE SHA256 LOOKUP
# ------------------------------------------------------------

metadata = pd.read_csv(METADATA_FILE, dtype={"SHA256": str})

metadata["SHA256"] = (
    metadata["SHA256"].astype(str).str.lower().str.strip()
)

metadata = metadata.drop_duplicates("SHA256")

metadata_lookup = metadata.set_index("SHA256")["label"].to_dict()

print("Metadata samples:", len(metadata_lookup))
print("Ransomware:", sum(v == 1 for v in metadata_lookup.values()))
print("Benign:", sum(v == 0 for v in metadata_lookup.values()))

# ------------------------------------------------------------
# 3. REPORT CATEGORY DEFINITIONS
# ------------------------------------------------------------

BEHAVIOUR_CATEGORIES = [
    "accessed_registry_keys",
    "created_registry_keys",
    "accessed_files",
    "created_files",
    "deleted_files",
    "changed_files",
    "created_process",
    "killed_process",
    "processes_tree",
    "accessed_mutexes",
    "created_mutexes",
    "loaded_modules",
    "executed_commands",
    "critical_api_calls"
]

NETWORK_CATEGORIES = [
    "network_traffic_ips",
    "network_dns_lookups"
]

ALL_CATEGORIES = BEHAVIOUR_CATEGORIES + NETWORK_CATEGORIES

# ------------------------------------------------------------
# 4. GENERAL HELPERS
# ------------------------------------------------------------

def safe_list(report, key):
    value = report.get(key, [])
    return value if isinstance(value, list) else []


def clean_string(value):
    if value is None:
        return ""
    return str(value).strip()


def extract_ips(value):
    """Extract valid IPv4/IPv6 addresses from observed text."""
    found = set()

    for candidate in re.findall(
        r'(?<![A-Za-z0-9])(?:\d{1,3}\.){3}\d{1,3}(?!\d)',
        clean_string(value)
    ):
        try:
            found.add(str(ipaddress.ip_address(candidate)))
        except ValueError:
            pass

    return found


def extract_executable_basenames(value):
    """
    Extract executable names from observed process/command text.
    These matches will be labelled inferred, not causal.
    """
    text = clean_string(value).lower()

    matches = re.findall(
        r'([a-zA-Z0-9_.-]+\.(?:exe|dll|bat|cmd|ps1|com))',
        text,
        flags=re.IGNORECASE
    )

    return {os.path.basename(x).lower() for x in matches}


# ------------------------------------------------------------
# 5. EVIDENCE-LINKED GRAPH CONSTRUCTION
# ------------------------------------------------------------

def build_evidence_graph(report):

    G = nx.Graph()

    # Network traffic IP observations
    traffic_ips = set()

    for item in safe_list(report, "network_traffic_ips"):
        traffic_ips.update(extract_ips(item))

    # DNS observations: domain and resolved IP addresses
    dns_records = []

    for item in safe_list(report, "network_dns_lookups"):
        text = clean_string(item)

        if " - " in text:
            left, right = text.rsplit(" - ", 1)
            domain = right.strip().lower()
        else:
            left = text
            domain = ""

        resolved_ips = extract_ips(left)

        dns_records.append({
            "domain": domain,
            "ips": resolved_ips
        })

    # DNS-to-IP edge only when IP is also observed in traffic.
    for record in dns_records:
        domain = record["domain"]

        if not domain:
            continue

        for ip in record["ips"] & traffic_ips:
            dns_node = ("DNS", domain)
            ip_node = ("IP", ip)

            G.add_edge(
                dns_node,
                ip_node,
                evidence="dns_ip_overlap",
                confidence="observed_overlap"
            )

    # Process observations
    process_names = set()

    for category in ["created_process", "processes_tree"]:
        for item in safe_list(report, category):
            process_names.update(extract_executable_basenames(item))

    # Command observations
    command_names = set()

    for item in safe_list(report, "executed_commands"):
        command_names.update(extract_executable_basenames(item))

    # Basename matches are inferred associations.
    for name in process_names & command_names:
        process_node = ("PROCESS", name)
        command_node = ("COMMAND", name)

        G.add_edge(
            process_node,
            command_node,
            evidence="basename_match",
            confidence="inferred"
        )

    return G


# ------------------------------------------------------------
# 6. GRAPH METRICS
# ------------------------------------------------------------

def calculate_graph_features(G):

    result = {}

    result["graph_nodes"] = G.number_of_nodes()
    result["graph_edges"] = G.number_of_edges()
    result["graph_density"] = nx.density(G)

    if G.number_of_nodes() == 0:
        empty_features = [
            "graph_components",
            "graph_average_degree",
            "graph_max_degree",
            "graph_max_k_shell",
            "graph_mean_k_shell",
            "graph_max_eigenvector",
            "graph_mean_eigenvector",
            "graph_mean_krylov_1",
            "graph_mean_krylov_2",
            "graph_mean_krylov_3",
            "graph_max_krylov_1",
            "graph_max_krylov_2",
            "graph_max_krylov_3",
            "largest_component_nodes"
        ]

        result.update({key: 0.0 for key in empty_features})
        return result

    degrees = dict(G.degree())

    result["graph_average_degree"] = float(np.mean(list(degrees.values())))
    result["graph_max_degree"] = int(max(degrees.values()))

    # Connected components
    components = sorted(
        nx.connected_components(G),
        key=len,
        reverse=True
    )

    result["graph_components"] = len(components)
    result["largest_component_nodes"] = max(map(len, components))

    # K-shell / core number
    if G.number_of_edges() > 0:
        core = nx.core_number(G)
    else:
        core = {node: 0 for node in G.nodes()}

    result["graph_max_k_shell"] = int(max(core.values()))
    result["graph_mean_k_shell"] = float(np.mean(list(core.values())))

    # Eigenvector centrality calculated per connected component.
    eigen_scores = {node: 0.0 for node in G.nodes()}

    for nodes in components:
        nodes = list(nodes)
        sub = G.subgraph(nodes)

        if len(nodes) <= 1 or sub.number_of_edges() == 0:
            continue

        try:
            scores = nx.eigenvector_centrality(
                sub,
                max_iter=1000,
                tol=1e-8
            )
        except nx.PowerIterationFailedConvergence:
            try:
                scores = nx.eigenvector_centrality_numpy(sub)
            except Exception:
                scores = {node: 0.0 for node in nodes}

        eigen_scores.update(scores)

    result["graph_max_eigenvector"] = float(max(eigen_scores.values()))
    result["graph_mean_eigenvector"] = float(np.mean(list(eigen_scores.values())))

    # Krylov features: normalized adjacency powers per component.
    node_krylov = {
        node: [0.0, 0.0, 0.0] for node in G.nodes()
    }

    for nodes in components:
        nodes = list(nodes)
        sub = G.subgraph(nodes)
        n = len(nodes)

        if n <= 1 or sub.number_of_edges() == 0:
            continue

        A = nx.to_scipy_sparse_array(
            sub,
            nodelist=nodes,
            weight=None,
            dtype=float,
            format="csr"
        )

        A = csr_matrix(A)

        degree_array = np.asarray(A.sum(axis=1)).ravel()

        inv = np.zeros(n, dtype=float)
        valid = degree_array > 0
        inv[valid] = 1.0 / np.sqrt(degree_array[valid])

        D = csr_matrix(
            (inv, (np.arange(n), np.arange(n))),
            shape=(n, n)
        )

        A_norm = D @ A @ D

        b = np.ones(n, dtype=float) / np.sqrt(n)

        v1 = A_norm @ b
        v2 = A_norm @ v1
        v3 = A_norm @ v2

        for node, a, b2, c in zip(
            nodes,
            np.asarray(v1).ravel(),
            np.asarray(v2).ravel(),
            np.asarray(v3).ravel()
        ):
            node_krylov[node] = [float(a), float(b2), float(c)]

    values = np.array(list(node_krylov.values()), dtype=float)

    result["graph_mean_krylov_1"] = float(values[:, 0].mean())
    result["graph_mean_krylov_2"] = float(values[:, 1].mean())
    result["graph_mean_krylov_3"] = float(values[:, 2].mean())

    result["graph_max_krylov_1"] = float(values[:, 0].max())
    result["graph_max_krylov_2"] = float(values[:, 1].max())
    result["graph_max_krylov_3"] = float(values[:, 2].max())

    return result


# ------------------------------------------------------------
# 7. BEHAVIOURAL AND NETWORK FEATURES
# ------------------------------------------------------------

def extract_report_features(report, sha256, label):

    row = {
        "SHA256": sha256,
        "label": int(label)
    }

    # Counts and unique observed values
    for category in ALL_CATEGORIES:
        values = [
            clean_string(x)
            for x in safe_list(report, category)
            if clean_string(x)
        ]

        row[category + "_count"] = len(values)
        row[category + "_unique"] = len(set(values))

    # Network protocol and port features
    protocol_counts = Counter()
    ports = set()
    traffic_ips = set()

    for item in safe_list(report, "network_traffic_ips"):
        text = clean_string(item)
        parts = text.split(" - ")

        if parts:
            protocol = parts[0].strip().upper()
            if protocol in ["TCP", "UDP", "ICMP"]:
                protocol_counts[protocol] += 1

        ips = extract_ips(text)
        traffic_ips.update(ips)

        # Extract port from final field when available.
        if len(parts) >= 3:
            try:
                port = int(parts[-1].strip())
                if 0 <= port <= 65535:
                    ports.add(port)
            except ValueError:
                pass

    row["network_unique_ips"] = len(traffic_ips)
    row["network_unique_ports"] = len(ports)
    row["network_tcp_count"] = protocol_counts["TCP"]
    row["network_udp_count"] = protocol_counts["UDP"]
    row["network_icmp_count"] = protocol_counts["ICMP"]

    # Evidence-linked graph
    G = build_evidence_graph(report)
    row.update(calculate_graph_features(G))

    return row


# ------------------------------------------------------------
# 8. FIND AND MATCH REPORTS
# ------------------------------------------------------------

json_files = glob.glob(
    os.path.join(REPORT_DIR, "**", "*.json"),
    recursive=True
)

print("\nJSON files discovered:", len(json_files))

# Sort for reproducibility
json_files.sort()

# ------------------------------------------------------------
# 9. PROCESS REPORTS WITH CHECKPOINTS
# ------------------------------------------------------------

batch = []
errors = []

processed = 0
matched = 0
unmatched = 0

# Start fresh. Delete old output files to avoid accidental duplication.
for path in [OUTPUT_FILE, ERROR_FILE]:
    if os.path.exists(path):
        os.remove(path)

for report_path in tqdm(json_files, desc="Processing RanDS reports"):

    sha256 = os.path.splitext(os.path.basename(report_path))[0].lower()

    # Match report filename to metadata SHA256
    label = metadata_lookup.get(sha256)

    if label is None:
        unmatched += 1
        continue

    try:
        with open(report_path, "r", encoding="utf-8") as f:
            report = json.load(f)

        if not isinstance(report, dict):
            raise ValueError("JSON root is not a dictionary")

        row = extract_report_features(report, sha256, label)
        batch.append(row)

        matched += 1

    except Exception as exc:
        errors.append({
            "SHA256": sha256,
            "report_path": report_path,
            "error": str(exc)
        })

    processed += 1

    if len(batch) >= CHECKPOINT_SIZE:
        pd.DataFrame(batch).to_csv(
            OUTPUT_FILE,
            mode="a",
            index=False,
            header=not os.path.exists(OUTPUT_FILE)
        )
        batch.clear()

    if len(errors) >= CHECKPOINT_SIZE:
        pd.DataFrame(errors).to_csv(
            ERROR_FILE,
            mode="a",
            index=False,
            header=not os.path.exists(ERROR_FILE)
        )
        errors.clear()

# Save remaining rows
if batch:
    pd.DataFrame(batch).to_csv(
        OUTPUT_FILE,
        mode="a",
        index=False,
        header=not os.path.exists(OUTPUT_FILE)
    )

if errors:
    pd.DataFrame(errors).to_csv(
        ERROR_FILE,
        mode="a",
        index=False,
        header=not os.path.exists(ERROR_FILE)
    )

# ------------------------------------------------------------
# 10. FINAL SUMMARY
# ------------------------------------------------------------

print("\n========== EXTRACTION SUMMARY ==========")
print("JSON files discovered:", len(json_files))
print("Matched reports processed:", matched)
print("Unmatched reports:", unmatched)

if os.path.exists(ERROR_FILE):
    error_df = pd.read_csv(ERROR_FILE)
    print("Failed reports:", len(error_df))
else:
    print("Failed reports: 0")

if os.path.exists(OUTPUT_FILE):
    result_df = pd.read_csv(OUTPUT_FILE)

    print("\nFinal feature dataset:", result_df.shape)
    print("Unique SHA256:", result_df["SHA256"].nunique())

    print("\nClass distribution:")
    print(result_df["label"].value_counts())

    print("\nGraph feature coverage:")
    print(
        result_df[
            ["graph_nodes", "graph_edges", "graph_components"]
        ].notna().sum()
    )

    print("\nSaved:", OUTPUT_FILE)
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv("/content/RanDS_Validated_Input.csv")

N = df["graph_nodes"]
E = df["graph_edges"]

# Expected metrics for an undirected simple graph
expected_density = np.where(
    N > 1,
    2 * E / (N * (N - 1)),
    0
)

expected_average_degree = np.where(
    N > 0,
    2 * E / N,
    0
)

# Audit conditions
checks = {
    "negative_nodes": N < 0,
    "negative_edges": E < 0,
    "edges_exceed_possible": E > (N * (N - 1) / 2),
    "invalid_max_degree": df["graph_max_degree"] > (N - 1),
    "density_mismatch": ~np.isclose(
        df["graph_density"], expected_density, atol=0.001
    ),
    "average_degree_mismatch": ~np.isclose(
        df["graph_average_degree"], expected_average_degree, atol=0.01
    ),
    "fractional_components": ~np.isclose(
        df["graph_components"],
        np.round(df["graph_components"]),
        atol=0.001
    ),
    "components_exceed_nodes": df["graph_components"] > N,
    "negative_components": df["graph_components"] < 0,
    "eigenvector_above_one": df["graph_max_eigenvector"] > 1.000001
}

print("DETAILED GRAPH AUDIT")
print("=" * 60)

for name, condition in checks.items():
    print(f"{name}: {int(condition.sum())}")

# Save problematic rows separately
problem_mask = np.zeros(len(df), dtype=bool)

for condition in checks.values():
    # Safe handling of both Pandas Series and NumPy arrays
    arr = condition.to_numpy() if hasattr(condition, "to_numpy") else np.asarray(condition)
    problem_mask |= arr

audit_df = df.loc[problem_mask].copy()

audit_df.to_csv(
    "/content/RanDS_Graph_Problem_Rows.csv",
    index=False
)

print("\nTotal problematic rows:", len(audit_df))
print("Saved: /content/RanDS_Graph_Problem_Rows.csv")

print("\nFirst 15 problematic samples:")
print(
    audit_df[
        ["SHA256", "label", "graph_nodes", "graph_edges",
         "graph_density", "graph_components",
         "graph_average_degree", "graph_max_degree",
         "graph_max_eigenvector"]
    ].tail(15).to_string(index=False)
)
```

```python
import json
import re
import os

JSON_DIR = "/content/rands_dynamic/dataset"

sample_hash = (
    "21ae5eead6ec2cc1ae891111cccf5c0e"
    "85efe5591aaba8abdbdbd24fef6374b3"
)

sample_path = os.path.join(
    JSON_DIR,
    sample_hash[:2],
    sample_hash + ".json"
)

# Find the file if the folder structure differs
if not os.path.exists(sample_path):
    import glob
    matches = glob.glob(
        os.path.join(JSON_DIR, "**", sample_hash + ".json"),
        recursive=True
    )
    if matches:
        sample_path = matches[0]
    else:
        raise FileNotFoundError("Sample JSON not found")

with open(sample_path, "r", encoding="utf-8") as f:
    report = json.load(f)

# Extract IP addresses
ip_pattern = re.compile(
    r"\b(?:\d{1,3}\.){3}\d{1,3}\b"
)

traffic_ips = set()

for item in report.get("network_traffic_ips", []):
    traffic_ips.update(ip_pattern.findall(str(item)))

dns_ip_pairs = []

for item in report.get("network_dns_lookups", []):
    ips = set(ip_pattern.findall(str(item)))

    for ip in ips:
        if ip in traffic_ips:
            dns_ip_pairs.append(ip)

print("NETWORK EVIDENCE INSPECTION")
print("=" * 50)

print("Traffic IPs:", sorted(traffic_ips))
print("DNS/IP overlaps:", sorted(set(dns_ip_pairs)))

print("\nObserved overlap relationships:")
for ip in sorted(set(dns_ip_pairs)):
    print(f"DNS_IP_OVERLAP -> {ip}")

print("\nNote: An overlap indicates shared IP evidence,")
print("not proof of causality or malicious communication.")
```

```python
import os
import glob
import json
import pandas as pd
from collections import defaultdict

# --------------------------------------------------
# CONFIGURATION
# --------------------------------------------------

JSON_DIR = "/content/rands_dynamic/dataset"
CSV_FILE = "/content/RanDS_Validated_Input.csv"

TOP_K = 20

# --------------------------------------------------
# LOAD DATASET LABELS
# --------------------------------------------------

df = pd.read_csv(CSV_FILE)

# label 1 = ransomware (Column names are case-sensitive: 'SHA256')
ransomware_hashes = set(
    df.loc[df["label"] == 1, "SHA256"].astype(str).str.lower()
)

print("Ransomware samples:", len(ransomware_hashes))

# --------------------------------------------------
# BEHAVIOURAL NODE CATEGORIES
# --------------------------------------------------

categories = {
    "registry": [
        "accessed_registry_keys",
        "created_registry_keys"
    ],
    "file": [
        "accessed_files",
        "created_files",
        "deleted_files",
        "changed_files"
    ],
    "network": [
        "network_traffic_ips"
    ],
    "dns": [
        "network_dns_lookups"
    ],
    "process": [
        "created_process",
        "killed_process",
        "processes_tree"
    ],
    "mutex": [
        "accessed_mutexes",
        "created_mutexes"
    ],
    "module": [
        "loaded_modules"
    ],
    "command": [
        "executed_commands"
    ],
    "api": [
        "critical_api_calls"
    ]
}

# --------------------------------------------------
# COUNT NODE OCCURRENCES
# --------------------------------------------------

node_occurrences = defaultdict(int)
node_samples = defaultdict(set)

json_files = glob.glob(
    os.path.join(JSON_DIR, "**", "*.json"),
    recursive=True
)

processed = 0
failed = 0

for path in json_files:

    sha256 = os.path.basename(path).replace(".json", "").lower()

    if sha256 not in ransomware_hashes:
        continue

    try:
        with open(path, "r", encoding="utf-8") as f:
            report = json.load(f)

        processed += 1

        for category, fields in categories.items():

            for field in fields:

                values = report.get(field, [])

                if not isinstance(values, list):
                    continue

                # Count each node once per report for affected samples
                unique_nodes_in_report = set()

                for value in values:

                    if value is None:
                        continue

                    node = str(value).strip()

                    if not node:
                        continue

                    node_key = (category, node.lower())

                    node_occurrences[node_key] += 1
                    unique_nodes_in_report.add(node_key)

                for node_key in unique_nodes_in_report:
                    node_samples[node_key].add(sha256)

    except Exception as e:
        failed += 1

print("\nProcessing completed")
print("Processed ransomware reports:", processed)
print("Failed reports:", failed)

# --------------------------------------------------
# CREATE TOP-K TABLE
# --------------------------------------------------

results = []

for (category, node), count in node_occurrences.items():

    results.append({
        "node_category": category,
        "node": node,
        "affected_samples": len(node_samples[(category, node)]),
        "total_occurrences": count
    })

top_nodes = pd.DataFrame(results)

top_nodes = top_nodes.sort_values(
    by=["affected_samples", "total_occurrences"],
    ascending=False
)

top_k_nodes = top_nodes.head(TOP_K).reset_index(drop=True)

top_k_nodes.index += 1
top_k_nodes.index.name = "rank"

print("\nTOP", TOP_K, "RANSOMWARE-AFFECTED NODES")
print("=" * 80)

print(top_k_nodes.to_string())

# --------------------------------------------------
# EXPORT RESULTS
# --------------------------------------------------

top_nodes.to_csv(
    "/content/All_Ransomware_Affected_Nodes.csv",
    index=False
)

top_k_nodes.to_csv(
    "/content/Top_K_Ransomware_Affected_Nodes.csv"
    , index=True
)

print("\nFiles saved:")
print("/content/All_Ransomware_Affected_Nodes.csv")
print("/content/Top_K_Ransomware_Affected_Nodes.csv")
```
